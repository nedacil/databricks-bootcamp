##PySpark vs SQL

```python
%sql
select * from samples.bakehouse.sales_customers
limit 5
```

```python
df = spark.read.table("samples.bakehouse.sales_customers")
#display(tabla)
```

```python
df_selected = df.select("customerID", "first_name", "last_name", "email_address", "phone_number", "address", "city", "state", "country", "continent", "postal_zip_code", "gender")
```

```python
df_selected.display()
df_selected = df.select("*")
df_limited = df_selected.limit(5)
df_limited.display()
```

```python
%sql
select distinct country
from samples.bakehouse.sales_customers
```

```python
df_distinct = df_selected.select("country").distinct()
df_distinct.display()
```
